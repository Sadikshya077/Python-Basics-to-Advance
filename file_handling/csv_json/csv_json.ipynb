```python
def readCSVFile(file_name):
    with open(f'{file_name}.csv','r') as file:
        reader=csv.reader(file)
        for row in reader:
            print(row)
```

# Dictionary Data to CSV

```python
person = [
    {
        'name': 'Ram',
        'age':'30',
        'address':'kathmandu'
    },
    {
        'name':'Shyam',
        'age':'25',
        'address':'Jhapa'
    }
]
```

```python
with open('person_detail.csv','w') as file:
    columns=['name','age','address']
    writer = csv.DictWriter(file,fieldnames=columns)
    writer.writeheader()
    writer.writerows(person)
```

```python
with open('person_detail.csv','r') as file:
    reader=csv.DictReader(file)
    for row in reader:
        for key,val in row.items():
            print(f'{key}: {val}')
        print('\n')
```

# JSON file

```python
person = [
    {
        'name': 'Ram',
        'age':'30',
        'address':'kathmandu'
    },
    {
        'name':'Shyam',
        'age':'25',
        'address':'Jhapa'
    }
]
```

```python
'''
-dump -> to create
-load -> to read
'''
```

```python
import json
with open('person.json','w') as file:
    json.dump(person,file,indent=4)
```

```python
with open('person.json','r') as file:
    data=json.load(file)
    for i in data:
        for key,val in i.items():
            print(f'{key}: {val}')
        print('\n')
```

```python
new_person ={
        'name': 'Ram',
        'age':'30',
        'address':'kathmandu'
    }
```

```python
with open('person.json','w') as file:
    data.append(new_person)
    json.dump(data,file,indent=4)
```
