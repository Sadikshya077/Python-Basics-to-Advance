# Binary File

```python
'''
- rb -> read binary (copy)
- wb -> write binary (paste)
------------------------------
Images, PDF, Word, PPT, Excel, Database files...
'''
```

```python
# open(path,operation)
with open('C:/Users/acer/Pictures/Screenshots/Screenshot 2026-07-29 074422.png', 'rb') as file:
    image = file.read()

with open('dummy.png', 'wb') as file:
    file.write(image)

# same for pdf or word files.
```
