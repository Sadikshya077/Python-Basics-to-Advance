# Text file

# Create file

```python
# open(file_name, operation)
file=open('text_file.txt','w')
file.write('Hello World')
file.close()
```

```python
# another new version that doesn't need close
with open('new_text.txt','w') as file:
    file.write('New text.')
```

# Read file

```python
file=open('text_file.txt','r')
content=file.read()
print(content)
```

```python
with open('new_text.txt','r') as file:
    content=file.read()
print(content)
```

# Add content

```python
with open('text_file.txt','a') as file1:
    file1.write('\tAdded text')
```

```python
file_name=input('Enter file name: ')
content=input('Enter content you want to add in the file: ')
with open(file_name,'a') as file:
    file.write(f'\n\t{content}')
```
