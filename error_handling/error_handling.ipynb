```python
'''
Error Handling
------------------
1. Try -> Runnable code
2. Except -> just like catch in C++ (Error show)
3. Else -> print or return statement
4. Finally -> anything that we want to show either error or not.
5. Raise -> Custom Exception
'''
```

```python
try:
    a= int(input('Enter first number'))
    b= int(input('Enter second number'))

    c=a/b
    print(c)
except ValueError:
    print('Error: Only numbers are accepted by the input field')
except ZeroDivisionError as e:
    print(f'Error: {e}')
# except Exception as e:
#     print(f'Error: {e}')  ----> for global handling 
```

```python
try:
    a= int(input('Enter first number'))
    b= int(input('Enter second number'))

    c=a/b    
except ValueError:
    print('Error: Only numbers are accepted by the input field')

except ZeroDivisionError as e:
    print(f'Error: {e}')
else:
    print(c)
finally:
    print('Program ended successfully')
```

# Custom Exception

```python
class AgeError(Exception):
    pass
```

```python
def ageCheck(age):
    if age>=18:
        return 'Vote'
    else:
        raise AgeError(f'AgeError: {age} is not eligible to vote.')
```

```python
try: 
    ageCheck(12)
except AgeError as e:
    print(e)
```

```python
class PinError(Exception):
    pass
class BalanceInsufficientError(Exception):
    pass
class BalanceIncorrectError(Exception):
    pass
```

```python
pin=1234
balance=50000
def atm():
    for i in range (3):
        entered_pin=int(input("Enter pin: "))
        if pin==entered_pin:
            amount=int(input("Enter amount you want to withdraw: "))
            if amount<balance:
                if amount%1000==0:
                    print("Withdrawn amount: ",amount)
                    print("Remaining amount: ",balance-amount)
                else:
                    raise BalanceIncorrectError("BalanceIncorrectError: Amount must be multiple of 1000")
            else:
                raise BalanceInsufficientError("BalanceInsufficientError: Amount entered is more than your current balance")
        elif pin!=entered_pin:
            if i==2:
                print("Account Blocked!")
                break
            raise PinError("PinError: Invalid pin, enter again")
```

```python
try:
    atm()
except PinError as e:
    print(e)
except BalanceInsufficientError as e:
    print(e)
except BalanceIncorrectError as e:
    print(e)
except ValueError as e:
    print(e)
```
