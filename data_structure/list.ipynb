```python
# Index
'''
Index starts from 0.
Index can neither be created nor removed
'''
alphabets = ['a','b','c','d','e','f','g']
alphabets[4]
```

```python
for data in alphabets:
    print(data)
```

```python
# unpacking list
print(*alphabets)
```

```python
# enumerate -> prints both index and value
for index, value in enumerate(alphabets):
    print(f'{index}:{value}')
```

```python
multi_d
```

```python
for i in multi_d:
    print(i)
    for j in i:
        print(j)
```

```python
for i_index, i_value in enumerate(multi_d):
    print('\n')
    print(f"{i_index}: {i_value}")
    for j_index, j_value in enumerate(i_value):
        print(f"{j_index}: {j_value}")
```

```python
fruits=['Apple','Banana','Coconut']
```
