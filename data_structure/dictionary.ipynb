# Dictionary

```python
'''
key and value pair
'''
data = {
    'name':'Ram',
    'age':30,
    'address':'Kathmandu',
    'qualification':'Masters'
}
data
```

```python
personal_details ={
    'Name':['Ram','Shyam','Hari'],
    'Age':[21,22,23],
    'Address':['Kathmandu','Bhaktapur','Lalitpur']
}
personal_details
```

```python
# JSON (Javascript object notation) dictionary inside a list.
# we cannot store multiple values inside a single key in this case.
data1=[
    {
        'Name':'Ram',
        'Age':21,
        'Address':'Kathmandu'
    },
    {
        'Name':'Shyam',
        'Age':22,
        'Address':'Bhaktapur'
    },
    {
        'Name':'Hari',
        'Age':23,
        'Address':'Lalitpur'
    }
]
data1
```

```python
'''
1. keys() -> extracts keys only
2. values() -> extracts values only
3. items() -> extracts both keys and values
'''
```

```python
for i in data.keys():
    print(i)
```

```python
for i in data.values():
    print(i)
```

```python
for key,val in data.items():
    print(f'{key}: {val}')
```
