# IF

```python
marks= float(input('Enter your marks: '))
if marks >= 40:
    print('Pass')
else:
    print('Fail')
```

```python
temperature= float(input('Enter temperature: '))

if temperature >= 35:
    print('Its hot today.')
elif temperature >= 20 and temperature < 35:
    print('Its warm today.')
else:
    print('Its cold today.')
```

```python
marks= float(input('Enter your marks: '))

if marks >= 70 and marks <= 100:
    print('Grade: A')
elif marks >=60 and marks < 70:
    print('Grade: B')
elif marks >= 50 and marks <60:
    print('Grade: C')
elif marks >= 40 and marks <50:
    print('Grade: D')
elif marks >= 0 and marks < 40:
    print('Fail')
else:
    print("Invalid marks.")
```

```python
age=int(input("Enter your age"))
if age<5:
    print("Price: 5")
elif age>=5 and age<=17:
    print("price: 10")
else:
    print('Price: 15')
```

```python
income=float(input("Enter your annual income"))
if income<=250000:
    print("Tax: 0%")
    tax=0
elif income > 250000 and income <=500000:
    print("Tax: 5%")
    tax=5/100*income
elif income >500000 and income <=1000000:
    print("Tax: 20%")
    tax=20/100*income
else:
    print("Tax: 30%")
    tax=30/100*income
print("Annual income:",income)
print("Taxable amount",tax)
print("Remaining Income after tax",income-tax)
```

```python
pin=1234
balance=50000
for i in range (3):
    entered_pin=int(input("Enter pin: "))
    if pin==entered_pin:
        amount=int(input("Enter amount you want to withdraw: "))
        if amount<balance:
            if amount%1000==0:
                print("Withdrawn amount: ",amount)
                print("Remaining amount: ",balance-amount)
            else:
                print("Amount must be multiple of 1000")
        else:
            print("Amount entered is more than your current balance")
    elif pin!=entered_pin:
        if i==2:
            print("Account Blocked!")
            break
        print("Invalid pin, enter again")
```

```python
'''
Authenticator system
'''
username='abc@gmail.com'
password="abcd"
pwd=''
attempt=3
uname = input("Enter username: ")
if uname!=username:
    print("Incorrect username")
else:
    while pwd!=password:
        print(f"Remaining login attempts: {attempt}")
        pwd=input("Enter password: ")
        if uname!=username and pwd!=password:
            print("Invalid username or password")
        else:
            if uname==username:
                if pwd==password:
                    print("Login successful")
                else:
                    print("Incorrect password")
            else:
                print("User not found")
        attempt-=1
        if attempt==0:
            print("Account Blocked!")
            break
```
