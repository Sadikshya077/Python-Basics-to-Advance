# match case

```python
# Restaurant Order System
# 1. Breakfast
# 2. Lunch
# 3. Snacks
# 4. Dinner
# 5. Drinks
# 6. Exit
attempt =4
for i in range (4):
    print("Welcome to Restaurant Management System")
    print("1. Breakfast")
    print("2. Lunch")
    print("3. Snacks")
    print("4. Dinner")
    print("5. Drinks")
    print("6. Exit")

    print("Remaining order",attempt)
    attempt-=1
    choice = int (input("Enter your choice to order: "))
    match choice:
        case 1:
            print("You have ordered Breakfast")
            print("1.Tea\n2.Bread\n3.Sausage\n4.Egg")
            order=input("Enter what you want to order")
            match order:
                case 1:
                    print("You have ordered Tea")
                case 2:
                    print("You have ordered Bread")
                case 3:
                    print("You have ordered Sausage")
                case 4:
                    print("You have ordered Egg")
                case _:
                    print("Item does not exist")
        case 2:
            print("You have ordered Lunch")
        case 3:
            print("You have ordered Snacks")
        case 4:
            print("You have ordered Dinner")
        case 5:
            print("You have ordered Drinks")
        case 6:
            print("Thank you for choosing us!")
            break
        case _:  # This is default
            print("Invalid input given!")
else:
    print("Order limit!")
```

# Break and Continue

```python
'''
break: if we want to stop the program forcefully.
continue: if we want to skip any specific condition.
'''
for i in range(10):
    if i==5:
        break
    print(i)
```

```python
for i in range(1,11):
    if i==5:
        continue
    print(i)
```

# while loop

```python
# Guess number
number=7
while True:
    guess_number=int(input("Enter the number you want to guess: "))
    if guess_number==number:
        print("You have guessed correctly.")
        break
    else:
        print("Wrong guess, try again!")
```

# Income Tax Calculator

```python
income=int(input('Enter your annual income: '))
if income<=250000:
    tax=0
elif income>250000 and income <=500000:
    tax= 5/100*income
elif income >500000 and income<=1000000:
    tax=20/100*income
else:
    tax=30/100*income
print(f'Tax amount: {tax}')
print(f'Total amount: {income+tax}')
```

# Withdraw amount (ATM)

```python
balance=100000
amount=int(input('Enter the amount you want to withdraw: '))
if amount%1000==0:
    if amount<balance:
        print('Amount withdrawn')
        rem=balance-amount
        print(f'Withdrawn amount: {amount}\nRemaining Balance: {rem}')
    else:
        print('Amount must be less than your current balance.')
else:
    print('Amount must be multiple of 1000')
```
