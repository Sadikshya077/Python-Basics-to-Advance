# Single Inheritance

```python
'''
1. Parent Class
2. Child Class

Types of Inheritance
----------------------------
1. Single Inheritance -> One parent, One child
2. Multiple Inheritance -> 
3. Multi-level Inheritance

Rules
----------------
- Child class can inherit the properties of parent class.
'''
```

```python
class Customers:
    def __init__(self, customer_id, customer_name, customer_type, customer_address, customer_phone):
        self.customer_id= customer_id
        self.customer_name=customer_name
        self.customer_type=customer_type
        self.customer_address=customer_address
        self.customer_phone=customer_phone
    def customer_details(self):
        print('Customer Details: ')
        print(f'Customer id: {self.customer_id}')
        print(f'Customer name: {self.customer_name}')
        print(f'Customer type: {self.customer_type}')
        print(f'Customer Address: {self.customer_address}')
        print(f'Customer Phone: {self.customer_phone}')
```

```python
class Orders(Customers):
    def __init__(self,order_id,order_status, order_date, order_destination,
                customer_id,customer_name, customer_type, customer_address, customer_phone):
        super().__init__(customer_id,customer_name, customer_type, customer_address, customer_phone)
        self.order_id=order_id
        self.order_status=order_status
        self.order_date=order_date
        self.order_destination=order_destination
    def order_details(self):
        print('Order Details: ')
        print(f'Order id: {self.order_id}')
        print(f'Order status: {self.order_status}')
        print(f'Order data: {self.order_date}')
        print(f'Order destination: {self.order_destination}')
```

```python
o=Orders(1,'pending','7/26/2026','Baneshwor','101','Ram','anything','Kapan','9876543210')
```

```python
o.customer_details()
print('\n')
o.order_details()
```

```python
class Category:
    def __init__(self,category_id, category_name):
        self.category_id=category_id
        self.category_name=category_name
    def category_details(self):
        print(f'Category id: {self.category_id}')
        print(f'Category name: {self.category_name}')
```

```python
class Product(Category):
    def __init__(self,product_id,product_name,product_price,category_id,category_name):
        super().__init__(category_id,category_name)
        self.product_id=product_id
        self.product_name=product_name
        self.product_price=product_price
    def product_details(self):
        print(f'Product ID: {self.product_id}')
        print(f'Product Name: {self.product_name}')
        print(f'Product Price: {self.product_price}')
```

```python
p=Product(5,'pant',2000,100,'clothes')
```

```python
p.category_details()
print('\n')
p.product_details()
```

# Multi-level Inheritance

```python
class Customers:
    def __init__(self, customer_id, customer_name, customer_type, customer_address, customer_phone):
        self.customer_id= customer_id
        self.customer_name=customer_name
        self.customer_type=customer_type
        self.customer_address=customer_address
        self.customer_phone=customer_phone
    def customer_details(self):
        print('Customer Details: ')
        print(f'Customer id: {self.customer_id}')
        print(f'Customer name: {self.customer_name}')
        print(f'Customer type: {self.customer_type}')
        print(f'Customer Address: {self.customer_address}')
        print(f'Customer Phone: {self.customer_phone}')
```

```python
class Orders(Customers):
    def __init__(self,order_id,order_status, order_date, order_destination,
                customer_id,customer_name, customer_type, customer_address, customer_phone):
        super().__init__(customer_id,customer_name, customer_type, customer_address, customer_phone)
        self.order_id=order_id
        self.order_status=order_status
        self.order_date=order_date
        self.order_destination=order_destination
    def order_details(self):
        print('Order Details: ')
        print(f'Order id: {self.order_id}')
        print(f'Order status: {self.order_status}')
        print(f'Order data: {self.order_date}')
        print(f'Order destination: {self.order_destination}')
```

```python
class Items(Orders):
    def __init__(self,item_id,quantity,unit_price,discount,order_id,order_status, order_date, order_destination,
                customer_id,customer_name, customer_type, customer_address, customer_phone):
        super().__init__(order_id,order_status, order_date, order_destination,
                customer_id,customer_name, customer_type, customer_address, customer_phone)
        self.item_id=item_id
        self.quantity=quantity
        self.unit_price=unit_price
        self.discount=discount
    def item_details(self):
        print(f'Item ID: {self.item_id}')
        print(f'Quantity: {self.quantity}')
        print(f'Unit Price: {self.unit_price}')
        print(f'Discount: {self.discount}')
        print(f'Total amount: {(self.unit_price*self.quantity)-self.discount}')
```

```python
i=Items(101,20,10,5,1,'pending','7/26/2026','Baneshwor','101','Ram','anything','Kapan','9876543210')
```

```python
i.customer_details()
```

```python
i.order_details()
```

```python
i.item_details()
```

# Multiple Inheritance

```python
class Staff:
    def __init__(self,staff_id,staff_name,staff_position):
        self.staff_id=staff_id
        self.staff_name=staff_name
        self.staff_position=staff_position
    def staff_details(self):
        print(f'Staff ID: {self.staff_id}')
        print(f'Staff Name: {self.staff_name}')
        print(f'Staff Position: {self.staff_position}')
```

```python
class Orders(Customers,Staff):
    def __init__(self,order_id,order_status, 
                 customer_id,customer_name, customer_type, customer_address, customer_phone,
                staff_id,staff_name,staff_position):
        
        Customers.__init__(self,customer_id,customer_name, customer_type, customer_address, customer_phone)
        Staff.__init__(self,staff_id,staff_name,staff_position)
        self.order_id=order_id
        self.order_status=order_status
    def order_details(self):
        print(f'Order id: {self.order_id}')
        print(f'Order status: {self.order_status}')
```

```python
o=Orders(101,'pending','101','Ram','anything','Kapan','9876543210','ST101','ABC','anything')
```

```python
o.customer_details()
```

```python
o.order_details()
```

```python
o.staff_details()
```
