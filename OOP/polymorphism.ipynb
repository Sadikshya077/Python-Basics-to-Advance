# Polymorphism

```python
class Khalti:
    def pay(self,amount):
        return f'Rs.{amount} paid through khalti.'
class Esewa:
    def pay(self,amount):
        return f'Rs.{amount} paid through Esewa'
class Fonepay:
    def pay(self,amount):
        return f'Rs.{amount} paid through Fonepay'
```

```python
def payment(balance,obj):
    print(obj.pay(balance))
```

```python
payment(600,Khalti())
```
