```python
class Dog():
    species='Mammal'
    name='Rocky'
    def sound(self): # we have to use self to ensure that the function is of same class 
        # as we have multiple classes inside a single file in case for python.
        return 'Bark'
```

```python
d=Dog()
```

```python
print(d.species)
print(d.name)
```

```python
d.sound()
```

```python
'''
1. Local Variable -> get user value through arguments
2. Instance Variable -> initialize user sent value (self.var_name) needed for constructor.
3. Global Variable -> it is defined outside the class and can be accessed by any other classes as well.
'''
```

```python
store_name='ABC'
class Product():
    def __init__(self, product_id, product_name, product_price, product_quantity, product_desc):
        self.store_name= store_name
        self.product_id= product_id
        self.product_name=product_name
        self.product_price=product_price
        self.product_quantity=product_quantity
        self.product_desc=product_desc
        
    def product_details(self):
        print(f'Store name: {store_name}')
        print(f'Product id: {self.product_id}')
        print(f'Product name: {self.product_name}')
        print(f'Product price: {self.product_price}')
        print(f'Product quantity: {self.product_quantity}')
        print(f'Product description: {self.product_desc}')
```

```python
p=Product(2,'Charger',3000,2,'Good Charger')
```

```python
p.product_details()
```
