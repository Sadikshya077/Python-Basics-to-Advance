```python
 '''
- Public
- Private (__)
- Protected (_)
'''
```

```python
# The data are private now
class Customers:
    def __init__(self,customer_id,customer_name,customer_phone):
        self.__customer_id=customer_id
        self.__customer_name=customer_name
        self.__customer_phone=customer_phone
    def customer_details(self):
        print(f'Customer ID: {self.__customer_id}')
        print(f'Customer Name: {self.__customer_name}')
        print(f'Customer Phone: {self.__customer_phone}')
```

```python
class Orders(Customers):
    def __init__(self,order_id,order_status,customer_id,customer_name,customer_phone):
        super().__init__(customer_name,customer_phone)
        self.__order_id=order_id
        self.__order_status=order_status
    def order_details(self):
        print(f'Order ID: {self.__order_id}')
        print(f'Order Status: {self.__order_status}')
```

```python
o=Customers(101,'ABC','98765432')
o.customer_details()
```

```python
# Protected
class Customers:
    def __init__(self,customer_id,customer_name,customer_phone):
        self._customer_id=customer_id
        self._customer_name=customer_name
        self._customer_phone=customer_phone
    def customer_details(self):
        print(f'Customer ID: {self._customer_id}')
        print(f'Customer Name: {self._customer_name}')
        print(f'Customer Phone: {self._customer_phone}')
```

```python
class Orders(Customers):
    def __init__(self,order_id,order_status,customer_id,customer_name,customer_phone):
        super().__init__(customer_id,customer_name,customer_phone)
        self._order_id=order_id
        self._order_status=order_status
    def order_details(self):
        print(f'Order ID: {self._order_id}')
        print(f'Order Status: {self._order_status}')
```
