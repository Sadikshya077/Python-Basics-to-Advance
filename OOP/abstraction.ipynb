```python
# A parent class that defines what its child classes MUST do
```

```python
# Parent class inherits Abstract Base Class
from abc import ABC, abstractmethod
class Bank(ABC):
    def connection(self):
        return 'Connected to server.'
    @abstractmethod
    def userBalance(self,balance):
        pass
```

```python
class MobileBanking(Bank):
    def userDetails(self):
        return f'User Name: ABC'
    def userBalance(self,balance):
        return f'Rs.{balance}'
```

```python
mb=MobileBanking()
```

```python
mb.userBalance(600)
```
