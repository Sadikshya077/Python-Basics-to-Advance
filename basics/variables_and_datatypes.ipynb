# Boolean 

```python
a= 5
b= 10
c= 15

'''
AND -> If all conditions are true its output will be true
----
1 1 -> 1
1 0 -> 0
0 1 -> 0
0 0 -> 0
'''

(a < c) and (b < c) and (c < a)
```

```python
'''
OR -> atleast one condition needs to be true to get true output.

1 1 -> True
1 0 -> True
0 1 -> True
0 0 -> False
'''

(a > c) or (b > c)
```

```python
'''
NOT
-----
True -> False
False -> True
'''

a == c
```
