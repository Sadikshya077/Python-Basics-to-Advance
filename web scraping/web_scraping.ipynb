```python
from selenium import webdriver
from selenium.webdriver.common.by import By
import pandas as pd
import time
```

```python
website = 'https://merolagani.com/LatestMarket.aspx'
```

```python
driver = webdriver.Chrome()
```

```python
driver.get(website)

time.sleep(3)
```

```python
table = driver.find_element(By.TAG_NAME, 'table')

table_body = table.find_element(By.TAG_NAME, 'tbody')

table_rows = table_body.find_elements(By.TAG_NAME, 'tr')
```

```python
table_head = table.find_element(By.TAG_NAME, 'thead')

headers = table_head.find_elements(By.TAG_NAME, 'th')

column_names = []

for header in headers:
    column_names.append(header.text.strip())

print(column_names)
```

```python
data = []

for row in table_rows:

    columns = row.find_elements(By.TAG_NAME, 'td')

    row_data = []

    for column in columns:
        row_data.append(column.text.strip())

    if row_data:
        data.append(row_data)
```

```python
df = pd.DataFrame(data, columns=column_names)

df.head()
```
