# Default Argument

```python
def product_detail(product_name='Unknown', product_price=0, product_quantity=0,
                   product_description='No description'):
    print(product_name)
    print(product_price)
    print(product_quantity)
    print(product_description)
```

```python
product_detail()
```

```python
# issue in default argument is if any parameter in between is missing, it still gives the values serially.
product_detail('Mobile','Android')
```

```python
# to solve this we use keyword argument
```

# Keyword Argument

```python
product_detail(product_name='Mobile',product_description='Android',product_price=200000)
```

# All Argument

```python
def person(*args):
    print(args)
    for data in args:
        print(data)
```

```python
person('Ram',20,'Bachelors','Kathmandu','9876543210')
```

# All Keywords Arguments (**kw_args)

```python
def kw_person(**kw_persons):
    print(kw_persons)

    for key,val in kw_persons.items():
        print(f'{key}: {val}')
```

```python
kw_person(name='Ram',age=21)
```

# Lambda Function

```python
# normal function
def calculate(a,b):
    add=a+b
    print(add)
```

```python
calculate(3,5)
```

```python
# var = Lambda a,b : calculation
add = lambda a,b : print(a+b) # to return we don't have to type return , simply a+b returns the sum.
```

```python
add(5,3)
```

```python
# if-else in one line 
marks=50
'Pass' if marks >= 40 else 'Fail'
```

```python
# we can only do one line operation using lambda function.
grade_calculator = lambda marks: 'Pass' if marks >=40 else 'Fail'
```

```python
grade_calculator(40)
```

```python
odd_even = lambda : [f'{i}: Even' if i%2==0 else f'{i}: Odd' for i in range(1,21)]
```

```python
odd_even()
```

```python
values = lambda : [i for i in range(1,21)]
```

```python
values()
```

```python
# above code is similar to:
def func():
    li=[]
    for i in range(1,21):
        li.append(i)
    print(li)
func()
```

```python
prices=[100,250,75,400,120,310]
cheap_items=[]
expensive_items=[]
func1 = lambda : [cheap_items.append(i) if i<200 else expensive_items.append(i) for i in prices]
```

```python
func1()
print(cheap_items)
print(expensive_items)
```
