```python
import sqlite3
```

```python
conn = sqlite3.connect('Daraz.db')
```

```python
cursor=conn.cursor()
```

# Create Table

```python
try:
    cursor.execute('''
        CREATE Table Customers(
        	customer_id integer primary key autoincrement,
        	customer_name text not null,
        	customer_phone text not NULL,
        	customer_address text not null,
        	customer_email text not null
        );
    ''')
    print('Customer table created successfully')
except sqlite3.OperationalError as e:
    print(e)
```

```python
try:
    cursor.execute('''
        Create table staff(
            staff_id integer primary key autoincrement,
            staff_name text not null,
            job_position text not null,
            job_type text not null,
            salary integer not null
        );
    ''')
    print('Staff table created successfully')
except sqlite3.OperationalError as e:
    print(e)
```

```python
try:
    cursor.execute('''
       Create table orders(
           order_id integer primary key autoincrement,
           order_status text not null,
           order_date date not null,
           order_description text not null,
           order_destination text not null,
           customer_id integer,
           staff_id integer,
           foreign key (customer_id) references customers(customer_id),
           foreign key (staff_id) references staff(staff_id)
           );
    ''')
    print('Order table created successfully')
except sqlite3.OperationalError as e:
    print(e)
```

# Insert Data

```python
try:
    cursor.execute('''
    Insert into Customers (customer_name,customer_phone,
    customer_address,customer_email)
    values('sita','0987654321','kapan','sita@gmail.com')
    ''')
    conn.commit()
    print('Data inserted successfully')
except sqlite3.OperationalError as e:
    print(e)
```

# Update Data

```python
try:
    cursor.execute('''
    update customers set customer_name="Bill",
    					 customer_phone="9856543012",
    					 customer_address="Lalitpur",
    					 customer_email="bill@gmail.com"
    where customer_id=1;
    ''')
    conn.commit()
except sqlite3.OperationalError as e:
    print(e)
```

# Fetch Data

```python
try:
    cursor.execute('''
    select * from customers;
    ''')
    data=cursor.fetchall()
    for val in data:
        print(val)
except sqlite3.OperationalError as e:
    print(e)
```
